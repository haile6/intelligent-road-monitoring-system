```python
import glob
vids = glob.glob("/kaggle/input/**/*.mp4", recursive=True)
gps  = glob.glob("/kaggle/input/**/gps_track.csv", recursive=True)
print("videos:", vids)
print("gps:", gps)

VIDEO   = vids[0]                    # change the index if you have several videos
GPS_CSV = gps[0] if gps else None    # None = no GPS
```

```python
import cv2, sqlite3, math, shutil, datetime as dt
import pandas as pd

DAMAGE_MODEL = "/kaggle/working/alli_pothole_best.pt"
OUT, DB_PATH, EVENTS = "/kaggle/working/output.mp4", "/kaggle/working/road_damage.db", "/kaggle/working/events"
VEHICLES = {2: "car", 3: "motorcycle", 5: "bus", 7: "truck"}
CONF, CONFIRM = 0.25, 3            # confidence, frames needed before a damage counts
MERGE_M, MERGE_S = 15, 10          # same damage within 15 m and 10 s = one event

track = pd.read_csv(GPS_CSV) if GPS_CSV else None
def get_gps(sec):
    if track is None: return None, None
    i = (track.sec - sec).abs().idxmin()
    if abs(track.sec[i] - sec) > 5: return None, None
    return float(track.lat[i]), float(track.lon[i])

def metres(a, b, c, d):
    p = math.pi / 180
    return 6371000 * math.hypot((d - b) * p * math.cos((a + c) / 2 * p), (c - a) * p)

def severity(name, box, shape):
    r = ((box[2]-box[0]) * (box[3]-box[1])) / (shape[0] * shape[1])
    hi, mid = (0.04, 0.01) if name == "pothole" else (0.12, 0.04)
    return "high" if r > hi else "medium" if r > mid else "low"

if os.path.exists(DB_PATH): os.remove(DB_PATH)
shutil.rmtree(EVENTS, ignore_errors=True); os.makedirs(EVENTS)
dmg_m, veh_m = YOLO(DAMAGE_MODEL), YOLO("yolo11s.pt")

db = sqlite3.connect(DB_PATH)
db.execute("""CREATE TABLE events(id INTEGER PRIMARY KEY AUTOINCREMENT, damage_type TEXT,
    confidence REAL, latitude REAL, longitude REAL, timestamp TEXT, image_path TEXT,
    severity TEXT, vehicle_type TEXT)""")

cap = cv2.VideoCapture(VIDEO)
fps = cap.get(cv2.CAP_PROP_FPS) or 30
w, h = int(cap.get(3)), int(cap.get(4))
writer = cv2.VideoWriter(OUT, cv2.VideoWriter_fourcc(*"mp4v"), fps, (w, h))
start = dt.datetime.now()
seen, counted, recent, n, merged = {}, set(), [], 0, 0

while True:
    ok, frame = cap.read()
    if not ok: break
    n += 1; sec = n / fps
    veh = veh_m.predict(frame, conf=0.4, classes=list(VEHICLES), verbose=False)[0]
    dmg = dmg_m.track(frame, conf=CONF, persist=True, verbose=False)[0]

    vtype = None
    for b, c in zip(veh.boxes.xyxy.int().tolist(), veh.boxes.cls.int().tolist()):
        vtype = VEHICLES[c]
        cv2.rectangle(frame, (b[0], b[1]), (b[2], b[3]), (255, 150, 0), 2)
        cv2.putText(frame, vtype, (b[0], b[1]-6), 0, 0.7, (255, 150, 0), 2)

    if dmg.boxes.id is not None:
        for b, c, cf, tid in zip(dmg.boxes.xyxy.int().tolist(), dmg.boxes.cls.int().tolist(),
                                 dmg.boxes.conf.tolist(), dmg.boxes.id.int().tolist()):
            name = dmg_m.names[c]
            col = (0, 0, 255) if name == "pothole" else (0, 200, 255)
            cv2.rectangle(frame, (b[0], b[1]), (b[2], b[3]), col, 3)
            cv2.putText(frame, f"{name} {cf:.2f}", (b[0], b[1]-8), 0, 0.8, col, 2)
            seen[tid] = seen.get(tid, 0) + 1
            if seen[tid] < CONFIRM or tid in counted: continue
            counted.add(tid)

            lat, lon = get_gps(sec)
            if lat is not None and any(r["name"] == name and sec - r["sec"] <= MERGE_S and
                                       metres(lat, lon, r["lat"], r["lon"]) <= MERGE_M for r in recent):
                merged += 1; continue
            path = f"{EVENTS}/{name}_{tid}.jpg"
            cv2.imwrite(path, frame)
            db.execute("INSERT INTO events(damage_type,confidence,latitude,longitude,timestamp,image_path,"
                       "severity,vehicle_type) VALUES(?,?,?,?,?,?,?,?)",
                       (name, cf, lat, lon, (start + dt.timedelta(seconds=sec)).isoformat(" ", "seconds"),
                        path, severity(name, b, frame.shape), vtype))
            db.commit()
            if lat is not None: recent.append({"name": name, "lat": lat, "lon": lon, "sec": sec})
    writer.write(frame)
    if n % 300 == 0: print(f"frame {n} | events {len(counted) - merged}")

cap.release(); writer.release()
total = db.execute("SELECT COUNT(*) FROM events").fetchone()[0]
db.close()
print(f"DONE. frames {n} | events in database: {total} | merged duplicates: {merged}")
```
